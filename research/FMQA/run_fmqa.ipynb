```python
import subprocess
```

```python
# max_vals = [1.0, 0.1, 0.01, 0.001, 0.0001]
# step_pts = [0.2, 0.25, 0.3]

max_vals = [0.1, 0.01, 0.001, 0.0001]
step_pts = [0.3]

# max_vals.extend( [0.2, 0.4, 0.6, 0.8] )
# max_vals.extend( [0.02, 0.04, 0.06, 0.08] )
# max_vals.extend( [0.002, 0.004, 0.006, 0.008] )
# max_vals.extend( [0.0002, 0.0004, 0.0006, 0.0008] )
```

```python
program_id = 5001    # 起始program_id
processes = []       # 儲存所有子進程

for max_val in max_vals:
    for step_pt in step_pts:

        command = ["python3", "fmqa.py", f"{step_pt}", f"{max_val}", f"{program_id}"]
        process = subprocess.Popen(command,
                                   stdin=subprocess.DEVNULL,
                                   stdout=subprocess.DEVNULL,
                                   stderr=subprocess.DEVNULL)
        print(f"子進程 PID: {process.pid} 已啟動。")
        processes.append(process)

        program_id += 1

# 等待所有子進程結束
for p in processes:
    p.wait()
```
